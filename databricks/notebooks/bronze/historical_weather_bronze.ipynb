# Historical Weather Bronze Processing

Read raw Historical Weather data from the ADLS Landing zone, apply minimal transformations, and load it into the Bronze Delta table for the Urban Mobility platform.

```python
from pyspark.sql.functions import current_timestamp, to_timestamp, when
```

```python
storage_account = "dubmobdev002"
container = "landing"
dataset = "weather"
year = "2026"
month = "05"
```

```python
# Landing path
landing_path = (
    f"abfss://{container}@{storage_account}.dfs.core.windows.net/"
    f"{dataset}/year={year}/month={month}"
)

# Read Historical Weather JSON files
df = spark.read.json(landing_path)

display(df)
```

```python
# Write to Bronze Delta table
(
    df.write
      .format("delta")
      .mode("overwrite")
      .saveAsTable("urban_mobility.bronze.historical_weather")
)
```

```python
# Verify the Bronze table
display(spark.table("urban_mobility.bronze.historical_weather"))
```
