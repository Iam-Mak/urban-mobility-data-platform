# Unity Catalog Setup

Configure the Unity Catalog foundation for the Urban Mobility platform.

### Design
- Separate **landing** and **managed** storage to isolate raw ingestion from curated data.
- Use Unity Catalog for centralized governance and secure data access.
- Organize managed data using the Bronze, Silver, and Gold layers.

```sql
-- External locations

CREATE EXTERNAL LOCATION IF NOT EXISTS urban_mobility_landing
URL 'abfss://landing@dubmobdev002.dfs.core.windows.net/'
WITH (STORAGE CREDENTIAL urban_mobility_cred);

CREATE EXTERNAL LOCATION IF NOT EXISTS urban_mobility_managed
URL 'abfss://managed@dubmobdev002.dfs.core.windows.net/'
WITH (STORAGE CREDENTIAL urban_mobility_cred);
```

```sql
-- Catalog
CREATE CATALOG IF NOT EXISTS urban_mobility
MANAGED LOCATION 'abfss://managed@dubmobdev002.dfs.core.windows.net/';
```

```sql
-- Schemas
CREATE SCHEMA IF NOT EXISTS urban_mobility.bronze
MANAGED LOCATION 'abfss://managed@dubmobdev002.dfs.core.windows.net/bronze';

CREATE SCHEMA IF NOT EXISTS urban_mobility.silver
MANAGED LOCATION 'abfss://managed@dubmobdev002.dfs.core.windows.net/silver';

CREATE SCHEMA IF NOT EXISTS urban_mobility.gold
MANAGED LOCATION 'abfss://managed@dubmobdev002.dfs.core.windows.net/gold';
```
