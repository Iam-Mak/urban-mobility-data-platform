
## External Location

```python
%sql

CREATE EXTERNAL LOCATION urban_mobility_landing
URL 'abfss://landing@dubmobdev001.dfs.core.windows.net/'
WITH (STORAGE CREDENTIAL urban_mobility_cred);
```

```python
%sql

CREATE EXTERNAL LOCATION urban_mobility_managed
URL 'abfss://managed@dubmobdev001.dfs.core.windows.net/'
WITH (STORAGE CREDENTIAL urban_mobility_cred);
```

```python
catalog_name = "urban_mobility"

spark.sql(f"""
CREATE CATALOG IF NOT EXISTS {catalog_name}
MANAGED LOCATION 'abfss://managed@dubmobdev001.dfs.core.windows.net/'
""")
```

```python
schemas = [
    "bronze",
    "silver",
    "gold"
]

for schema in schemas:
    spark.sql(f"""
    CREATE SCHEMA IF NOT EXISTS {catalog_name}.{schema}
    """)
```
